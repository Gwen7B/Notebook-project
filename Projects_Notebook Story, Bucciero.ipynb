# Chapter 4: Break Even Analysis
Below is some quick math and explanation of how many units a month Orbiters will have to sell to stay profitable. The information is sourced from an external document with more detailed calculations and notes. 

There are three values we need to calculate the break-even point: The monthly fixed costs, and the average contribution margin per stuffed animal  

The average contribution margin is calculated by multiplying the contribution (retail price - manufacturing cost) by the percentage of inventory each product occupies and adding all of the values together. 

| Product | Quantity | Share of Inventory (in percentage) |
| ------- | ----- | ------------- |
| Classic | 200 | 55.6% |
| Mini | 120 | 33.3% |
| Hefty | 40 | 11.1% |

```python
(0.3333*15.53)+(0.5556*22.62)+(0.1111*25.28)
```

- Monthly fixed costs: $2,300
- Average contribution margin: $20.55

To calculate the break-even units, we use the following formula:

Break-even units = Monthly Fixed Costs / Contribution per Unit

```python
2300/20.55
```

This means that each month Orbiters must sell **approximately 112 units** to be profitable.
